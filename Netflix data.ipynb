```python
# Step 1: Import Required Libraries

import numpy as np 
import pandas as pd
```

```python
# Step 2: Load the Dataset

data = pd.read_excel("C:/Raw data files/Netflix Dummy Data/Netflix_Unclean_Project.xlsx")
print(data)
```

```python
# step 3: Explore the dataset
# view first 5 records

data.head()
```

```python
# Check the columns

data.columns
```

```python
# view Dataset Information

data.info()
```

```python
# Change the dtype (datetime64[us])  

data['Join_Date'] = data['Join_Date'].astype('datetime64[ns]')
data['Last_Watch_Date'] = data['Last_Watch_Date'].astype('datetime64[ns]')
```

```python
# Statistical summary

data.describe(include = "number")
```

```python
# Step 4: Check Missing Values

data.isnull().sum()
```

```python
# Step 5: Check Duplicate Records

data.duplicated().sum()
```

```python
# Remove duplicates Records

data.drop_duplicates(inplace = True)
```

```python
# Check again duplicates records

data.duplicated().sum()
```

```python
# Step 6: Dirty Values ("N/A", "NULL", " ", Space)
# Replace common invalid values with NaN

data.replace(["N/A", "NULL", " "], np.nan, inplace = True)
```

```python
# Check again Missing Values

data.isnull().sum()
```

```python
# Remove Extra Space

data = data.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
# Step 7: Standardize Text Format
# Convert text into Proper Case.

data["Country"] = data["Country"].str.title()
data["Genre"] = data["Genre"].str.title()
data["Subscription_Plan"] = data["Subscription_Plan"].str.title()
data["Account_Status"] = data["Account_Status"].str.title()
```

```python
# Step 8: Convert Data Columns

data["Join_Date"] = pd.to_datetime(data["Join_Date"], errors = "coerce")
data["Last_Watch_Date"] = pd.to_datetime(data["Last_Watch_Date"], errors = "coerce")
```

```python
# Step 9: Convert Numeric Columns

numeric_cols = ["Age", "Watch_Hours", "Monthly_Fee", "Movies_Watched",
               "Episodes_Watched", "Avg_Rating", "Monthly_Sessions", 
                "Download_Count"]
```

```python
for c in numeric_cols:
    data[c] = pd.to_numeric(data[c], errors = "coerce")
```

```python
# Check again Statistical summary

data.describe(include = "number")
```

```python
# Correct Age column data

data = data[(data["Age"]>=18) & (data["Age"]<=100)]
```

```python
# Check Age columns data

data["Age"].describe()
```

```python
# Negative Values

data = data[data["Watch_Hours"]>=0]
data = data[data["Monthly_Fee"]>=0]
```

```python
# Date Cleaning 

data["Join_Date"] = pd.to_datetime(data["Join_Date"], errors = "coerce")
data["Last_Watch_Date"] = pd.to_datetime(data["Last_Watch_Date"], errors = "coerce")
```

```python
data[["Join_Date", "Last_Watch_Date"]].isnull().sum()
```

```python
# Step 10: Filling Missing Values

data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Watch_Hours"] = data["Watch_Hours"].fillna(data["Watch_Hours"].mean())
data["Monthly_Fee"] = data["Monthly_Fee"].fillna(data["Monthly_Fee"].mean())
```

```python
# Categorical Columns

data["Payment_Method"] = data["Payment_Method"].fillna("Unknown")
data["Device"] = data["Device"].fillna("Unknown")
data["Genre"] = data["Genre"].fillna("Unknown")
```

```python
# Step 11: Feature Engineering
# Customer Revenue

data["Customer_Revenue"] = (data["Monthly_Fee"] * data["Monthly_Sessions"])
print(data)
```

```python
# Engagement Score

data["Engagement_Score"] = data["Watch_Hours"] + data["Monthly_Sessions"] + data["Download_Count"]
print(data)
```

```python
# Watch Category

bins = [0, 20 , 50, 100, np.inf]
label = ["Low","Medium","High","Very High"]

data["Watch_Category"] = pd.cut(data["Watch_Hours"], bins = bins, labels = label)
print(data)
```

```python
# Watch Category Count All

data["Watch_Category"].value_counts()
```

```python
# Customer Age Group

bins = [17,25,35,50,100]
label = ["18-25","26-35","36-50","50+"]

data["Age_Group"] = pd.cut(data["Age"], bins = bins, labels = label)
print(data)
```

```python
# Age Group Count All

data["Age_Group"].value_counts()
```

```python
# Check Account is active or not

data["Account_Flag"] = np.where(data["Account_Status"] == "Active", 1,0)
print(data)
```

```python
# Account Active Count All

data["Account_Flag"].value_counts()
```
